```python
import pandas as pd
from datetime import datetime

def to_dd_mm_yyyy(date_str):
    """
    Convert a date string to 'dd/mm/yyyy' format.
    Supports:
      - 'dd/mm/yyyy' (e.g., '06/03/2009')
      - 'yyyy-mm-dd' (e.g., '1990-02-10')
    Returns the original string if format is unknown or if input is not a string.
    """
    if not isinstance(date_str, str):
        return date_str  # keep NaN / None / numbers as they are

    # Remove any leading/trailing spaces
    date_str = date_str.strip()

    # Try dd/mm/yyyy first (dayfirst=True)
    try:
        dt = datetime.strptime(date_str, "%d/%m/%Y")
        return dt.strftime("%d/%m/%Y")
    except ValueError:
        pass

    # yyyy-mm-dd
    try:
        dt = datetime.strptime(date_str, "%Y-%m-%d")
        return dt.strftime("%d/%m/%Y")
    except ValueError:
        pass

df['published_date'] = df['published_date'].apply(to_dd_mm_yyyy)
```

```python
# filling empty scraped time
df["scraped_at"] = df["scraped_at"].fillna(pd.Timestamp.now(tz="UTC"))
# convert those into the right data type
df["duration_min"] = pd.to_numeric(df["duration_min"], errors="coerce").astype("Int64")
df["year_of_birth"] = pd.to_numeric(df["year_of_birth"], errors="coerce").astype("Int64")
df["part"] = pd.to_numeric(df["part"], errors="coerce").astype("Int64")

# fill null parts into zeros (as those has 0 parts)
df["part"] = df["part"].fillna(0).astype(int)
df["duration_min"] = df["duration_min"].fillna(0).astype(int)

# fill null speakers and gender as unknown instaed of null
df["speaker"] = df["speaker"].fillna("unknown")
df["gender"] = df["gender"].fillna("unknown")

# the rest is filled with nulls as those values can not be filled with
#anything else
df = df.where(pd.notnull(df), pd.NA)
```

```python
import pandas as pd
from datetime import datetime

# The goal is to split long text entries into overlapping chunks of ≤512 words
# (or tokens) while keeping the associated metadata.

def chunk_text(text, chunk_size=512, overlap=50):
    """
    Split text into overlapping word chunks.

    Args:
        text (str): Input text
        chunk_size (int): Maximum words per chunk
        overlap (int): Number of overlapping words between consecutive chunks

    Returns:
        list: List of text chunks (strings)
    """
    words = text.split()
    if not words:
        return []

    step = chunk_size - overlap
    if step <= 0:
        raise ValueError("overlap must be less than chunk_size")

    chunks = []
    for i in range(0, len(words), step):
        chunk_words = words[i:i + chunk_size]
        # Only keep chunks longer than 5 words
        if len(chunk_words) > 5:
            chunks.append(" ".join(chunk_words))

    return chunks

def expand_long_texts(df, text_col='text', chunk_size=512, overlap=50):
    """
    Expand DataFrame: each chunk of a long text becomes a separate row.
    Updates the word_count column for each chunk.

    Args:
        df (pd.DataFrame): Input DataFrame
        text_col (str): Name of the column containing text
        chunk_size (int): Maximum words per chunk
        overlap (int): Overlap between chunks in words

    Returns:
        pd.DataFrame: Expanded DataFrame with one row per chunk
    """
    new_rows = []

    for idx, row in df.iterrows():
        text = row[text_col]

        # Skip if text is NaN or empty
        if not isinstance(text, str) or len(text.strip()) == 0:
            continue

        chunks = chunk_text(text, chunk_size, overlap)

        if not chunks:
            # No chunking needed (text already short enough)
            # Keep original row as is, but ensure 'chunk_index' is set to 0
            new_row = row.to_dict()
            new_row['chunk_index'] = 0  # Assign 0 for non-chunked rows
            new_rows.append(new_row)
        else:
            # Create one row per chunk
            original_word_count = row.get('word_count', len(text.split()))
            for i, chunk in enumerate(chunks):
                new_row = row.to_dict()
                new_row[text_col] = chunk
                new_row['word_count'] = len(chunk.split())
                new_row['chunk_index'] = i
                new_rows.append(new_row)

    return pd.DataFrame(new_rows)

# Expand
df_expanded = expand_long_texts(df, chunk_size=512, overlap=50)

df_expanded["chunk_index"] = df_expanded["chunk_index"].astype(int)
df_expanded["word_count"] = df_expanded["word_count"].astype(int)

# Check result
print(f"Original rows: {len(df)}  ->  Expanded rows: {len(df_expanded)}")
df_expanded.head()
```
